#Section 3: Variables

Next, let's introduce another fundamental yet incredibly important part of python: Variables

Think of variables like a box. The way that variables work is that they store data, so that you're able to access the data you want by making reference to the variable name!

Variable names can be almost anything, from single letters like 'a' or 'x' to really long sentences without spacings like 'thisismysupersupersuperlongvariablename', but there are some restrictions:
1. Variable names cannot start with a number, only a letter or an underscore, '_'.
2. Variable names can only contain alpha-numeric characters (A-Z, 0-9, so special characters like @!#%^$& or even spaces are not permitted)
3. Variable names are case sensitive, so two variables named 'age' and 'AGE' will be treated as two completely different variables.
4. Variables cannot be any of the python keywords(e.g. print, input)

Values are assigned to variables using this format 'variablename = data'

#Section 4: Data Types

Today, we'll only be covering 4 data types:

- Strings: basically sentences like 'hi my name is xxx' or 'the quick brown fox jumps over the lazy dog'. The use of quotation marks is specifically for strings.

- Integers: whole numbers like 1, 7, 23.

- Float: also numbers, but float specifically allows for decimal points to be used, so 13.2, 72.134

- Boolean(will be important later on in functions and loops): only True or False (yes the captial letter is very important)

If you are unsure, you can use print out the data type of whatever it is by using 'print(type(datayouwanttocheck))'.

```python
#string, str for short
print(type('hello this is actually a string'))

#integer, int for short
print(type(1))

#float
print(type(1.1111111))

#boolean, bool for short
print(type(True))
```

#Section 5: Basic Arithmetic

Did you know that you could do math in python? Well now you do!

Here is a list of the symbols used and what they translate to in math:

'+' : addition

'-' : subtraction

'*' : multiplication

'/' : division

'**': exponentiation (raising something to the power of something)

*//*: floor function, gives you the quotion

'%' : modulus, gives you the remainder

Try and predict the output of each of the print() functions to test your knowledge :D

```python
print(1 + 1)
print(1 - 1)
print(2 * 2)
print(3 / 2)
print(3 ** 2)
print(5 // 2)
print(5 % 2)
```
