## Section 1: Lists

In programming, we often need to store **multiple values**.

Instead of creating many variables:

`

<br>
`

We can store everything in **one structure** called a list.

Think of a list like a container holding multiple items.

````python
# Creating a list
fruits = ["apple", "banana", "cherry"]

print(fruits)
````

Each item in a list has a position called an **index**.

IMPORTANT:
- Index starts from 0 (not 1!)

````python
print(fruits[0])  # first item
````

### Working with Lists (Operations)

Now that we know how to create a list, let’s walk through some common ways to **manipulate lists**:

### ➕ Adding items

- `append()` → adds **one item** to the end  
- `extend()` → adds **multiple items** from another list  
- `insert()` → adds an item at a **specific index (position)**  
- `+` (list addition) → combines two lists into a **new list**

### ➖ Removing items

- `remove()` → removes an item by its **value**  
- `pop()` → removes an item by its **position** (or last item by default)

### 🔍 Exploring the list

- `len()` → tells you how many items are in the list  
- `index()` → finds the position of a value  
- `in` → checks if an item exists in the list  

### 🔄 Organising the list

- `sort()` → sorts the original list  
- `sorted()` → creates a new sorted list (keeps original unchanged)


Now, let’s try these operations in code.

````python
fruits = ["apple", "banana", "cherry"]
print("Starting list:", fruits)

# ADDING ITEMS

fruits.append("orange")
print("After append:", fruits)

fruits.extend(["grape", "melon"])
print("After extend:", fruits)
#to the back of the list

fruits.insert(1, "kiwi")  # position 1
print("After insert:", fruits)

new_list = fruits + ["mango", "pear"]
print("Using + operator (new list):", new_list)

# REMOVING ITEMS

fruits.remove("banana")
print("After remove:", fruits)

popped_item = fruits.pop() #only removes the last element
print("Popped item:", popped_item)
print("After pop:", fruits)

# ACCESSING & CHECKING

print("Length of list:", len(fruits)) #how many items are there in the list

print("Index of apple:", fruits.index("apple")) #position

print("Is 'apple' in list?", "Apple" in fruits)

# SORTING

sorted_list = sorted(fruits)
print("Sorted (new list):", sorted_list)
print("Original list still:", fruits)

fruits.sort()
print("After sort (in-place):", fruits)

# STRING TO LIST

sentence = "apple banana orange"
words = sentence.split()

print("Split sentence into list:", words)
````

## Section 2: Loops with Lists

Sometimes, we want to go through every item in a list. Instead of accessing each index of the array like Array[n], we can use for and while loops within lists!

````python
numbers = [1, 2, 3, 4, 5] #length = 5

print("Using for loop:")
for n in numbers:
    print(n)

print("Using while loop:")
i = 0
while i < len(numbers):
    print(numbers[i])
    i += 1

#while loop terminates when i == 5
````

## Section 3: Tuples

At first glance, tuples look almost identical to lists.

BUTTTT they aren't the same!!!

A **tuple is a list that cannot be changed** after it is created.

We call this **immutability**.

Tuples are immutable, lists are mutable.


### Why does this matter?

Sometimes, we want data to stay *unchanged always*:
- coordinates (x, y)
- dates (day, month, year)
- settings that should not be accidentally modified (eg: personal information)

Tuples are perfect for this!

Let's look at how to manipulate tuples.

````python
# Creating a tuple
coordinates = (10, 20)
print("Tuple:", coordinates)

# ACCESSING (same as lists)
print("First value:", coordinates[0])

print("Length:", len(coordinates))

print("Is 10 in tuple?", 10 in coordinates)

# LOOPS
for value in coordinates:
    print("Value:", value)

# Tuples cannot be changed
#coordinates[0] = 50  # This will cause an error (comment this line out to ensure successful run of code)


# CONVERSION
# list to tuple
numbers_list = [1, 2, 3]
numbers_tuple = tuple(numbers_list)
print("Converted to tuple:", numbers_tuple)

# tuple to list
back_to_list = list(numbers_tuple)
back_to_list.append(4) #adding
print("Modified list:", back_to_list)
````

## Section 4: Dictionaries

Lists use positions (index numbers) to store data.

But sometimes, that’s not very intuitive.

For example, would you rather remember:
- student[0]
or
- student["name"] ?

It will be easier to remember descriptive words rather than pure numbers, so this is where you can use dictionaries!


### What is a Dictionary?

A dictionary stores data in **key-value pairs**. Unlike Lists and Tuples, Dictionaries are have no numerical index. Instead, they have keys to replace the index numbers.

Example:
<br>
`
"name" : "Alex"  
`
<br>
`
"age" : 16  
`
<br>
Each key-value pair is separated by a colon `:`

Note: keys in dictionaries are Case Sensitive!

### Why use dictionaries?

They make data:
- easier to understand  
- easier to access  
- more structured  

````python
# Creating a dictionary
student = {
    "name": "Alex",
    "age": 16,
    "subjects": ["Math", "Science"]
}

print("Original:", student)

# ACCESSING
print("Name:", student["name"])

# MODIFYING / ADDING
student["age"] = 17 #change value
student["grade"] = "A++" #add new keys
student["subjects"].append("English") #modify list inside the dictionary

print("After updates:", student)

# REMOVING
student.pop("age")
print("After removing age:", student)

# CHECKING
print("Number of keys:", len(student))
print("Is 'name' a key?", "name" in student)

# MERGING
extra_info = {"school": "ABC High"} #created another dictionary
student.update(extra_info)

print("After merging:", student)

# LOOPING
for key, value in student.items():
    print(key, ":", value)
````

## Section 5: Functions

As your code gets longer and longer and longer... you tend to repeat code.
<br>
Here's where you can use fucntions to prevent long lengthy code.

### What is a Function?

A function is a **reusable block of code** that performs a task.

Instead of rewriting the same code again and again, you define the function once, and reuse it whenever needed.

### Why use functions?

- makes code cleaner
- avoids repetition
- easier to debug and understand  

This is how a function goes:
<br>
input --> process --> output

````python
# create a function
def greet(name):
    return "Hello " + name

print(greet("Crystal")) #prints the return value of Hello Crystal
````

This is how a basic function syntax will look like. Take note that a **parameter** is a placeholder variable in the function, while the **argument** is the actual value passed in when calling your function.
<br></br>
Let's look at what other possible functions there can be!

````python
#addition
def add(a, b): #2 parameters!!
    return a + b

print(add(3, 5)) #2 arguments!!

print("i added!")

# this function checks if a number is even
def check_even(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(check_even(5))

print("i evened")

# LOCAL VARIABLES (SCOPE)
def example(): #has 0 parameters!
    x = 10  # local variable
    print("Inside function:", x)

example() #has 0 arguments!

x = 20
print(x) #This will cause an error (x is local, you cant access a local variable!!)

print("i can local and global")

# FUNCTIONS WITH LISTS
def print_list(items):
    for item in items:
        print(item)

print_list(["apple", "banana", "cherry"])
````
