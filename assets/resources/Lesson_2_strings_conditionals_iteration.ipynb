## 2. String Slicing Operations

### (i) Using `[start:stop]`

We can use `[start:stop]` to perform string slicing operations:
- `start` denotes the beginning index of the substring. The substring will **include** the character at the `start` index, or we can say it is **inclusive**.
- `stop` denotes the ending index of the substring. The substring will **exclude** the character at the `stop` index, or we can say it is **exclusive**.

````python
s = "Hello World!"
print(s[0:3])
print(s[4:7])
````

### Note:
If the `start` index value is missing, it will take the default beginning index as '0'. If the `stop` index value is missing, the default ending index is the 'length of the string'.

For example:

````python
s = "Hello World!"
print(s[:3])
print(s[7:])
print(s[:])
````

### Strings are Immutable

Once a string is created, it is impossible to change its content.


````python
s = "Hello World!"
# Uncomment the following line to see what would happen when we try to change one character inside the string.
# s[0] = "h" - This is not allowed
s = "h" + s[1:]
print(s)
````

However, we can use string slicing and concatenation to create new strings.  
**Take note that since strings are immutable, `new_s` is totally a new string object and has no relationship with the original string `s`.**

````python
# If we want to change the captital letters "H" and "W" to lowercase, we use the following approach:
s = "Hello World!"
new_s = "h" + s[1:6] + "w" + s[7:]
s = new_s
print(new_s, s)
````

### Using the `find()` method

We can also use the built-in method `str.find()` to find out the starting index of the **first** occurrence of a substring.

If a subtring is not found, the value `-1` will be returned instead of raising a `ValueError`.

````python
s = "Hello World"
print(s.find("haha"))   # substring not found. returns -1
print(s.find("o"))
````

### Note:

`index()` method can also be used in other sequence objects (iterable objects) such as `list` or `tuple`. Whereas `find()` method is ONLY applicable for `string` objects.

````python
lst = [1,2,3,4,5,6]
print(lst.index(3))
print(lst.find(3))
````

### Using the `isdigit()` and `isalpha()` methods

We can use the built-in method `str.isdigit()` and `str.isalpha()` to determine if a string value contains digits or alphabets.

````python
print("9".isdigit())
print("z".isalpha())
print("55505".isdigit())
print("aa".isalpha())
print("a6a".isdigit())
print("P_P".isalpha())     # underscore present
print("9.78".isdigit())    ## decimal point present
````

### Using the `ord()` and `chr()` functions

We can use the built-in function `ord()` and `chr()` to convert from character to its Unicode 97 values and from Unicode 97 values back to character.

`ord()` - Ordinal of one character to Unicode 97    
`chr()` - Unicode 97 to character      

**Take note that the string parameter used in `ord()` must have a length of 1; and integer parameter in `chr()` must be valid unicode representation.**

````python
char = 99
print(ord("1"))
print(chr(char))
````

### Functions manipulating upper case and lower case
(i) We can use the built-in methods `str.isupper()` and `str.islower()` to check if the character is an upper case or lower case.    
(ii) We can also use built-in methods `str.upper()` and `str.lower()` to **convert** to uppercase and lowercase respectively.

````python
strg1 = 'abcd'
strg2 = 'ABCD'

# str.isupper()
print(strg1.isupper())
print(strg1.islower())

# str.islower()
print(strg2.isupper())
print(strg2.islower())

# Converts to uppercase and lowercase using lower() and upper()
print(strg1.upper())
print(strg2.lower())
print('12ab!@c4'.upper())
````

## 4. String Formatter
`str.format()` is one of the string formatting methods in Python 3, which allows multiple substitutions and value formatting. This method lets us concatenate elements within a string through positional formatting.

### Using a String Formatter
Formatters work by putting in one or more replacement fields and placeholders defined by a pair of curly braces { } into a string and calling the `str.format()`. The value we wish to put into the placeholders and concatenate with the string arguments passed as parameters into the format function.
Exception error will occur when there are extra placeholders with no parameters

````python
fruit1 = "durian"
print("I love {}!".format(fruit1))
````

````python
fruit2 = input("Key in favourite fruit: ")
print("I love {}!".format(fruit2))
````

````python
fruit3 = "apple"
print("I love {}!".format(fruit3))
````

````python
# Multiple placeholders
fruit_4 = "apple"
fruit_5 = "orange"
print("I love {} and {}!".format(fruit_4,fruit_5))
````

````python
# Default positional arguments is 0, 1, 2 ...
print("I love {} and {}!".format(fruit_4,fruit_5))
print("I love {0} and {1}!".format(fruit_4,fruit_5))

# What if index is swapped?
print("I love {1} and {0}!".format(fruit_4,fruit_5))
````

````python
# New Arguments
print("Now I love {new_fruit} and {0}!".format(fruit_4,fruit_5, new_fruit = "durian"))
````

````python
# Note: 2 arguments as parameters with 3 placeholders
print("I love {0}, {1} and {0}!".format(fruit_4,fruit_5))
````

### Note: An alternative to `.format()` formatting
With newer versions of Python, `f"string{arg}"` is released as a replacement for `"str{}".format(arg)` method.

````python
fruit1 = "durian"
print(f"I love {fruit1}!")
````

What happens if there are more arguments as parameters than placeholders?

````python
fruit_3 = "pear"
print("I love {} and {}!".format(fruit_4, fruit_5, fruit_5))
print("I love {} and {}!".format(fruit_4, fruit_5, fruit_6))
````

`elif` is equivalent to `else if`.
  
Here is an example:

Using nested `else` + `if`

````python
temperature = 32

if temperature > 30:
    print("It is a hot day")
else:
  if temperature < 24:
    print("It is a cold day")
  else:
    print('It is a normal day')
````

## Section 1 - `while` loop

### 1.1 Syntax of `while` loop

Loop is one of the most important programming construct which enables us to repeatedly perform a set of determined action to a set of data.

We shall start to learn and understand how `while` loops operate first. Here is a basic syntax of a `while` loop.

```python
while conditional_expression:
    perform_certain_actions
```

When the loop is executed, the `conditional_expression` will be evaluated. If the condition is `True`, the program will run for **one iteration** before evaluating the `conditional_expression` again.

The loop will then be repeated until `conditional_expression` is evaluated to be `False`, then it will stop.

````python
x = 0   # intitalise indexing variable

while x <= 3:   # while true/false?
    x += 1      # increment by 1
    print(x)

print("End")
````

### **Control Flow Tools: `break` and `continue`**

We can further control the loops by using the `break` and `continue` statements.

`break` statement will stop the current iteration and skip all the rest of the iterations.

**Predict the outcome of the following two programs. What's the difference and why?**

````python
x = 0
while x < 10:
    print(x, end = " ")
    if x == 5:
        break
    x += 1
print("end loop")
````

````python
x = 0
while x < 10:
    if x == 5:
        break
    print(x)
    x += 1
print("end loop")
````

`continue` statement will stop the current iteration and continue with the next round of iteration.

````python
x = 0
while x < 10:
    x += 1
    if x == 5:
        continue
    print(x)
print("end loop")
````

### 2.1 `for` loop to iterate using items in a collection

### **`range()` function**

The `range()` function returns a sequence of numbers, starting from 0 by default, and increments by 1 (by default), and ends at a specified number.

Syntax of `range()` funcion:

```python
range(start, stop, step)
```

Similar to string slicing operations, `range()` function also takes in `start`, `stop` and `step` values.

````python
# range with only "stop" value
print('range with only "stop" value (5)')
for i in range(5):
    print(i)
````

````python
# range with "start" and "stop" value
print('range with "start" (2) and "stop" value (5)')
for i in range(2, 5):
    print(i)
````

````python
# range with "start", "stop" and "step" value
print('range with "start" (2), "stop" (5) and "step" (2) value')
for i in range(2, 5, 2):
    print(i)
````

````python
print('range with "start" (0), "stop" (10) and "step" (2) value')
for i in range(0, 10, 2):
    print(i)
````

### **`for` loop to iterate using index**

Combining with the `range()` function, we can create for loops to iterate using the index values of a sequence.

````python
str1 = "Hello World!"

for i in range(len(str1)):
    print(str1[i])
````

### **Comparison between `for` loop and `while` loop**

A `while` loops is more flexible, it can be used to repeat a specific block of code an **unknown** number of times, until a condition is met.

On the other hand, to make use of `for` loops, we **must** know exactly how many times to execute the particular section of code.

````python
num = 0
while num <= 10:
    print(num, end = ' ')
    num += 1
````

````python
for num in range(10):
    print(num, end = ' ')
````

## **Nested Loops**

> Add blockquote



A nested loop is a loop within a loop, an inner loop within the body of an outer loop.


````python
outer = 0
while outer < 3:
    inner = 0
    print("outer: ", outer)
    while inner < 3:
        print(inner, end = " ")
        inner += 1
    outer += 1
    print()
````

````python
for outer in range(3):
    print("outer: ", outer)
    for inner in range(3):
        print(inner, end = " ")
    print()
````
