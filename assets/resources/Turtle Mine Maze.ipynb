## Optional: More about recursion

When do we use recursive functions?

1. When solution of a problem depends on solutions to smaller instances of the same problem
  * fractals
  * Euclidean algorithm

2. When we are trying to search through a list of possibilities.
  * Searching thorugh tree-like structures

In fact, Recursive functions achieve the same outcome as loops: Anything that can be done with loops can be done with recursive functions. Same goes the other way round.

Each have their own advantages.

### Factorial

For example, a function that finds the factorial of a number

A factorial is represented by an exclamation mark, and e.g.

5! = 5 x 4 x 3 x 2 x 1

> Or more generally: n = n x (n-1) x (n-2) x ... x 2 x 1

Each factorial function always ends with 1, and we can define that as the base case

>A base case in recursion defines the stopping condition for the recursive function

factorial(3)

= 3 x factorial(2)

= 3 x 2 x factorial(1)

= 3 x 2 x 1

= 6

```
def factorial(n):
  if n == 1: # our base case
    break
  else:
    return(n*factorial(n-1))
```

A visualisation:

The program first runs through each of the recursive case, until it reaches the base case. Then, it returns each value, up the ladder (shown in the yellow blocks in the image), to obtain the final answer.

###File directory Search

Files in our devices have a structure that resembles this:

Where files are categorised and nested under folders.

image.png


If we want to print all file names, we can use a recursive function that calls itself when it meets a folder.

In pseudocode,

```
def list_files(folder):
    for item in folder:
        if item is a folder:
            list_files(item)
        else:
            print(item)
```


This function does the following (Using the same directory as above):

```
list_files(Documents)

  list_files(School)

      print Math.pdf

      print Notes.txt

  list_files(Photos)

      list_files(Holiday)
```
