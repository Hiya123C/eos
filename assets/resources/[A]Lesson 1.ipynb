Exercise 1: Do some simple calculations using 2 inputs from a user and print the result!

Hint: as mathematical operations can only be done on integers/floats, you can use 'int(input('input))' or 'float(input('input))' to convert the user input to an integer or float!

```python
#user input 1
number1 = float(input('enter a number: '))
#user input 2
number2 = float(input('enter a second number: '))
#addition, assign result to a variable
sum = number1 + number2

#subtraction, assign result to a variable
subtract = number1 - number2

#multiplication, assign result to a variable
product = number1 * number2

#division, assign result to a variable
divided  = number1 / number2

#print all 4 new variables
print(f'the sum of number 1 and 2 is {sum}!')
print(f'the result when number 2 is subtracted from number 1 is {subtract}!')
print(f'the product of number 1 and 2 is {product}!')
print(f'the result when number 1 is divided by number 2 is {divided}!')
```

Exercise 2: Get the user's name, age and favourite colour. Print a greeting, how old they will be in 5 years and their favourite colour.

Remember how to join variables and strings together?

```python
#user input - user's name
user_name = input('enter your name: ')

#user input - user's age
user_age = int(input('enter your age: '))

#user input - user's favourite colour
user_favourite_colour = input('enter your favourite colour: ')

#print - greet the user with their name
print(f'hello {user_name}!')

#print - age of the user in 5 years
in_5_years = user_age + 5
print(f'your age in 5 years is {in_5_years}!')

#print - user's favourite colour
print(f'your favourite colour is {user_favourite_colour}!')
```
