### **Challenge Exercise 3**
Check whether a year is a leap year.

Rules:
- Divisible by 400 -> leap year
- Divisible by 100 -> not leap year
- Divisible by 4 -> leap year
- Otherwise -> not leap year

````python
year = int(input('Enter year: '))

if year % 400 == 0:
    is_leap_year = True
elif year % 100 == 0:
    is_leap_year = False
elif year % 4 == 0:
    is_leap_year = True
else:
    is_leap_year = False

if is_leap_year:
  print(f"{year} is a leap year!")
else:
  print(f"{year} is not a leap year!")
````

### **Optional Exercise**

In mathematics, the binomial coefficients are the positive integers that occur as coefficients in the binomial theorem. Commonly, a binomial coefficient is indexed by a pair of integers n ≥ r ≥ 0 and is written (n,r) It is the coefficient of the x^r term in the polynomial expansion of the binomial power (1 + x)^n, and it is given by the formula:  
&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;n!/(r!(n-r)!)

n! is n factorial where n! = n(n-1)(n-2)..(2)(1)

Example: 5! is (5)(4)(3)(2)(1)

note that 0! = 1 and 1! = 1

Take in n and r which are 2 integers, the exponent **n** for the (1+x)^n expansion and the exponent **r** for the x^r term. Print the integer binomial coefficient with the exponent r term.

**Example**

Sample output for `n = 5` and `r = 2`  

Output: `10`

````python
# Write your codes here.

n = int(input())
r = int(input())

ans = 1
# calculate n! / r!
for i in range(r+1, n+1):
  ans *= i

for i in range(1, n-r+1):
  ans //= i

print(ans)
````

### **Challenge Exercise 8**

Given a number, `n`, display a pattern as follows:

Eg. `n = 5` will produce the following:

```
    1
   121
  12321
 1234321
123454321
```

````python
n = int(input('Enter n: '))

# Write your codes here.
for i in range(n):
  line = ''
  for j in range(i+1):
    line += str(j+1)
  for j in range(i, 0, -1):
    line += str(j)

  x = n-1-i
  line = x*' ' + line
  print(line)
````
