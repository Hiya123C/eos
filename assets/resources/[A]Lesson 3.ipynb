# Assignment Time!

The dictionary `menu_price` and `menu_name` is given in the format
`{item_index:item_price}` and `{item_index:"item_name"}`,
and the list `orders` is given in the format
`[("customer_id", item_id, ordered_quantity), ...]`.

Write 3 functions that allow the user to:

- `print_menu()`: Print the entire menu with the index, name and price.

**Expected Output:**

```
-------------Menu-------------
1 Cheesecake              70.0
2 Chocolate cake          15.0
3 Straweberry Shortcake   25.3
4 Apple pie               20.0
5 fish pie                10.0
6 Macarons                35.5
```

- `add_order(customer_id, item_id, ordered_quantity)`: Add an order to the `orders` list in the specified format and print the updated list.

**Expected Output:**

```
[('A002', 1, 12), ('C109', 3, 29), ('H772', 3, 10), ('Z992', 3, 10)]
```

- `expenditure(customer_id)`: Calculate the total expenditure of a customer given the customer ID. If the customer ID does not exist in the list, return a printed error message "Customer order not found."

**Expected Output:**

```
840.0
253.0
Customer order not found.
```

````python
#run this cell before testing your code!
menu_name = {
    1: "Cheesecake",
    2: "Chocolate cake",
    3: "Straweberry Shortcake",
    4: "Apple pie",
    5: "fish pie",
    6: "Macarons"
}

menu_price = {
    1: 70.00,
    2: 15.00,
    3: 25.30,
    4: 20.00,
    5: 10.00,
    6: 35.50
}

orders = [
    ("A002", 1, 12),
    ("C109", 3, 29),
    ("H772", 3, 10)
]
````

````python
#1st function
def print_menu():
    returned = "-------------Menu-------------\n"

    for idee, name in menu_name.items():
        returned += "{:<2}".format(str(idee))
        returned += "{:<24}".format(name)
        returned += "{:>3}".format(str(menu_price[idee]))
        returned += "\n"

    print(returned)

# Executable code (Do not remove)
print_menu()
````

````python
#2nd function
def add_order(customer_id, item_id, ordered_quantity):
    orders.append((customer_id, item_id, ordered_quantity))
    print(orders)

# Executable code (Do not remove)
add_order("Z992", 3, 10)
````

````python
#3rd function
def expenditure(customer_id):
    for tup in orders:
        if tup[0] == customer_id:
            print(menu_price[tup[1]] * tup[2])
            return

    print("Customer order not found.")

# Executable code (Do not remove)
expenditure("A002")
expenditure("Z992")
expenditure("hahaha")
````
