### What is File?

File is a named location on disk to store related information.

A file operation takes place in following order:
* Open a file

### Open and Close File as a habit

Python has a built-in function `open()` to open a file.
* The `open()` function, also called a handler, returns a file object. It then can be used to read or modify the file accordingly.
* After usage, always close the opened file using the `close()` method of the file object.

```python
f = open(filepath)
f.close()
```

### Filepath
The `filepath` can be a relative or absolute path.

#### Absolute Path
* An absolute path is a complete path to a file or directory from the root directory. The root directory is the top-level directory of the file system and is represented by a forward slash (/). Absolute paths always start with the root directory and provide the full path to the file or directory.

* For example, the absolute path to the home directory of a user named "Peter" would be "/home/Peter".

#### Relative Path
* A relative path is a path to a file or directory that is relative to the current directory. It specifies the location of the file or directory in relation to the current directory. Relative paths do not start with the root directory and are usually shorter than absolute paths.

* For example, if you are currently in the home directory of "Peter" and want to access a file named "example.txt" in a subdirectory called "documents", then the relative path would be "documents/example.txt".

The main difference between absolute and relative paths is how they are interpreted. Absolute paths always point to the same location regardless of the current directory, whereas relative paths may point to different locations depending on the current directory.

#### Note:
* If only file name is specified, Python assume the file is in the same folder as Python executable.

Python's open() function supports several modes that determine how a file is opened and manipulated. Here are the most common modes:

'r' (Read mode):
Default mode if none is specified.
Opens the file for reading.
The file must exist; otherwise, a FileNotFoundError is raised.

'w' (Write mode):
Opens the file for writing.
Creates the file if it doesn't exist.
Overwrites the file if it does exist.

'a' (Append mode):
Opens the file for writing.
Creates the file if it doesn't exist.
Appends new data to the end of the file without deleting its existing content.

'x' (Exclusive creation mode):
Creates a new file for writing.
Fails if the file already exists, raising a FileExistsError.

'b' (Binary mode):
Opens the file in binary mode.
Used for reading or writing binary data (e.g., images, videos).
Often combined with other modes like 'rb' or 'wb'.

't' (Text mode):
Opens the file in text mode.
This is the default mode for reading or writing text files.
Often combined with other modes like 'rt' or 'wt'.

'+' (Read and write mode):
Opens the file for both reading and writing.

Can be combined with other modes like 'r+', 'w+', 'a+', 'b+', etc.
'r+': Reads and writes, but file must exist.
'w+': Writes and reads, and file is created or truncated.
'a+': Reads and writes, with all writes appended to the end of the file.
Examples:
'rb': Open a file for reading in binary mode.
'wb': Open a file for writing in binary mode, overwriting existing content.
'r+': Open a file for both reading and writing.
'a+': Open a file for reading and appending.

````python
# Creates a text and Write to file
f1 = open('test2.txt', 'a+') # Append, read and write mode
f1.write("stu\n")
print(f1.read())
f1.close()

f2 = open('test.txt', 'r')
print(f2.read())
f2.close()
````

### Impt: Using Context Manager

A context manager in Python is an object that is used to manage resources, such as file handling, network connections, and database connections. It provides a way to allocate and release resources automatically when they are no longer needed, ensuring proper cleanup and preventing resource leaks.

Context managers are typically used with the `with` statement, which guarantees that the resources managed by the context manager are properly released at the end of the block, even if an exception occurs.

Example:  
`
with open('test.txt', 'w') as file:
    file.write('This is test.')`

With context manager, file.close() will be called automatically when with code block exits. Meaning the file will be closed after executing the block.

````python
with open('test.txt', 'a') as file:
    file.write('Hello Singapore ')

with open('test.txt') as file:
    print(file.read())
````

### Mode

You can specify the mode used to open a file by applying a second argument to the open function.
* `r`/`w`/`a`: Are you reading (r), writing (w) or appending (a) a file?
* `t`/`b`: Is it a text (t) or binary (b) file?

```python
with open(filename,mode) as file:
```

#### Read / Write / Append
The `mode` specifies how you want to work with the file.
* "r": read mode, which is the default.
* "w": write mode, for **overwriting** the contents of a file. Existing file content will be lost.
* "a": append mode, for appending new content to the end of the file. Existing content in the file will not be lost.

````python
## Using Context Manager

# This removes a file if it exists
import os
try:
    os.remove("hello.txt")
except:
    pass

# write/create a file
with open("hello.txt", "w") as file:
    file.write("Hello World\n")

# append/create a file
with open("hello.txt", "a") as file:
    file.write("This line was appended.\n")

# read a file
with open("hello.txt", "r") as file:
    content = file.read()
    print(content)
````
