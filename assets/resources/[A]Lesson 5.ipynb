# What is OOP?
>  A style of programming that involves lumping data and actions into objects that can be acted on like any normal object.

> no need to make a long list of instructions when you can lump it into one object



Take for example a car. It has
  1) attributes (data, eg "red, toyota, max speed")
  2) actions (behaviour, eg "accelerate, brake, turn, move")

In python, we can create an *object* (in this case the car) from a *class*

Classes are blueprints/template for objects. They define the **structure** and **behavior** of objects.

Creating a new object is called `instantiation`. An **object** of a class is also called **instance** of that class. An `instance` is a particular object created from a given class.





###Example 1.1

Lets look at an example below involving the behaviours and attributes of a dog

```python
##we create a class *Dog* by using *class*
## def __init__() initialises the attributes and behaviours of the dog

class Dog:
    def __init__(self,name,age,colour,breed,hungry):
        self.name = name
        self.age = age
        self.colour = colour
        self.breed = breed
        self.hungry = hungry

##functions within classes are known as methods
    def is_hungry(self):
        return self.hungry ##returns the behaviour

    def bark(self):
        print(self.name + ' barks: woof woof!')

    def get_breed(self):
        return self.breed

my_dog = Dog('wang wang', 4, 'white', 'labrador', True) ##input the attributes and behaviours
ur_dog = Dog('patrol', 3, 'brown', 'german shephard', False)

print(my_dog.bark())
print(ur_dog.is_hungry())
print(my_dog.is_hungry())
print(ur_dog.get_breed())
```

##2. What is '___str_ __' in OOP?

> Python provides a string method `__str__()` to convert an object to string. The str() method returns the "informal" or nicely printable representation of a given object, which is suitable to present information to end-user.



###Example 2.1

```python
class dog5:
    def __init__(self, name, hage, colour, breed):
        self.name = name
        self.hage = hage
        self.colour = colour
        self.breed = breed
        self.hungry = False #if not present will be created for you
        self.owner = []

    def set_owner(self,owner1):
        self.owner.append(owner1)

    def get_owner(self):
        return self.owner[-1]

    def get_hage(self):
        return self.hage

    def dog_years(self):
        dage = self.get_hage()*7  ##simple example off using method within method!
        return dage

    def __str__(self):
        strg = ''
        strg += f'Name: {self.name}.\n'
        strg += f'Breed: {self.breed}.\n'
        strg += f'Human Age: {self.hage}. \n'
        strg += f'Dog Age: {self.dog_years()}. \n'
        if len(self.owner) != 0:
            strg += f'Owner: {self.get_owner()}'
        else:
            strg += 'Owner: no owner.'

        return strg
```

### 2.1 Encapsulation - Accessors (getter) and Mutators (setter)

**Encapsulation** refers to the bundling of data with the methods that operate on that data. Public methods such as getters and setters access or modify the data and other classes call these methods for accessing.

> By **hiding variables** inside a class, it prevents public program codes to interfere/modify with the private variables within the class directly.

When a class attribute is named with double-leading-underscore, it invokes **name mangling**.

You will notice that for the rest of the examples, we will slightly modify the attribute names to make them less exploitable. Eg. *self._name* instead of the easily guessable *self.name*

##3. Inheritance and Polymorphism

> Sounds scary but dont worry -- it is quite simple!

> This chapter is about using classes within classses. Let's say we first create a class A. Later on, we want to use the methods from A inside another class, B.

>If class A is being used inside class B, it is known as the _superclass_ (parent class) and class B is known as the _subclass_ (child)

```python
##let's start off looking at these two classes

class Student:
    def __init__(self, name, class_):
        self._name = name
        self._class = class_

    def get_name(self):
        return self._name

s = Student("Xiao Ming", "3I")
print(s.get_name())

class Teacher:
    def __init__(self, name, classes):
        self._name = name
        self._classes = classes

    def get_name(self):
        return self._name

t = Teacher("Mr T.T Sahur", ["3C", "3I"])
print(t.get_name())
```

####notice anything similar about these two classes?

if you guessed that they both use the same _name_ attribute and the same method _get_name_, well done!

since repeating these methods and attributes is repetitive, let us create a superclasss (parent) that can provide the _name_ and _get_name_

```python
# here is the superclass, Person
class Person:
    def __init__(self, name):
        self._name = name

    def get_name(self):
        return self._name


class Student(Person):  # Student becomes the subclass of Person class
    def __init__(self, name, class_):
        self._name = name
        self._class = class_  ##compared to earlier, we do not have to rewrite the get_name method


class Teacher(Person):  # Teacher becomes the subclass of Person class
    def __init__(self, name, classes):
        self._name = name
        self._classes = classes

p = Person("buns")
print(p.get_name())

s = Student("patty", "4D")
print(s.get_name())

t = Teacher("Mr lettuce", ["3C", "3I"])
print(t.get_name())

```

Create a Python class `Point` which contains 2 attributes, `x` and `y`, representing x and y coordinate of the point.
* Implement initializer method which initialize `x` and `y`.
* Implement a instance method `distToOrigin()` which return distance from origin using formula `math.sqrt(x**2 + y**2)`.
* Implement `__str__()` which returns "(x,y)", e.g. "(3.0,4.0)"




```python
class Point:
    def __init__(self, x, y):
        self._x = x
        self._y = y
    def __str__(self):
        strg =  f'({self._x}, {self._y})'
        return strg #better to have extra step

    def distToOrigin(self):
        import math
        distance = math.sqrt((self._x)**2 + (self._y)**2)
        return round(distance,1)



### Test
p = Point(3.5,4.7)
print(p)
print(p.distToOrigin())
```

* implement a function `longest_length()` which has a parameter to take in a list argument containing Point objects. The function will return the Point with the longest distance from origin with the message print out: "Longest distance from origin is Point (x,y): (distance) units".

```python
def longest_length(points_list):
    lengthlist = []
    for pt in points_list:
        length = pt.distToOrigin()
        lengthlist.append(length)
    lengthlist.sort()
    dist = lengthlist[-1]
    for pt in points_list:
        if pt.distToOrigin() == dist:
            coord = pt.__str__()


    return f'Longest distance from origin is Point {coord}: {dist}'


## Test
a, b, c, d, e, f, g = Point(5.0,8.0), Point(7.0,4.0), Point(10.0,2.0), Point(4.0,11.0), Point(11.0,6.0), Point(2.0,15.0), Point(4.0,4.0)
points_list = [a,b,c,d,e,f,g]
print(longest_length(points_list))
```
