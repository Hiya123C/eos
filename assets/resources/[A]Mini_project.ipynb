````python
import random

def generate_mine():
  num_mines = 100
  mines_coord = []
  for i in range(num_mines):
    mines_coord.append((random.randint(-150,150), random.randint(-150,150)))
  return mines_coord #return in order to refer to these values later

def draw_mines(mines_coord):
  for i in mines_coord: #iterate through every mine tuple in the list
    mines.jumpto(i)
    mines.pendown()
    mines.dot()
    mines.penup() #to prevent lines between each dot


#for testing the fns
#mine = generate_mine()
#draw_mines(mine)
#mines.hideturtle()
````

````python
def check_mine(coord): #should return true if theres collision
  for i in mine_coord:
    if abs(coord[0] - i[0]) <= 5 and abs(coord[1] - i[1]) <= 5:
    # distance between x cor of turtle and mine < 5
    # coord[0] refers to the next x coord the turtle will move to
    # coord[1] refers to the next y coord the turtle will move to

    #It is highly unlikely that a mine will just nice happen to be in the path of the turtle, so we include a range
      return True
      break
  return False
````

To explain this:
```
heading = (90, 0, 180, 270)
change_in_coord = ((0, 5),(5, 0),(-5,0),(0,-5))
```
Index refers to index in the heading and change_in_coord tuple
```
Index | Heading | Delta  | Direction
0     | 90deg   | (0,+5) | Up
1     | 0deg    | (+5,0) | Right
2     | 180deg  | (-5,0) | Left
3     | 270deg  | (0,-5) | Down
```



````python
#for loop function

def move():
  for i in range(4): # it tries to move 5 pixels in every direction in ORDER
  # ORDER: up, right, left, down

    #up, right, left, down
    heading = (90, 0, 180, 270)
    change_in_coord = ((0, 5),(5, 0),(-5,0),(0,-5))
    delta = change_in_coord[i]
    dir = heading[i]
    #using setheading instead of left/right for absolute dirns, not relative
    bob.setheading(dir) # rotates bob to face that direction
    target_coord = (  # where bob will end up
      bob.xcor() + delta[0],
      bob.ycor() + delta[1]
    )
    # for debugging: print(check_mine(target_coord))
    if check_mine(target_coord) == False: # if no mine
      bob.forward(5)
      break



````
