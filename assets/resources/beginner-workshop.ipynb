#Section 4: Variables

Next, let's introduce another fundamental yet incredibly important part of python: Variables

Think of variables like a box. The way that variables work is that they store data, so that you're able to access the data you want by making reference to the variable name!

Variable names can be almost anything, from single letters like 'a' or 'x' to really long sentences without spacings like 'thisismysupersupersuperlongvariablename', but there are some restrictions:
1. Variable names cannot start with a number, only a letter or an underscore, '_'.
2. Variable names can only contain alpha-numeric characters (A-Z, 0-9, so special characters like @!#%^$& or even spaces are not permitted)
3. Variable names are case sensitive, so two variables named 'age' and 'AGE' will be treated as two completely different variables.
4. Variables cannot be any of the python keywords(e.g. print, input)

Values are assigned to variables using this format 'variablename = data'

## Section 6: Functions

As your code gets longer and longer and longer... you tend to repeat code.
<br>
Here's where you can use fucntions to prevent long lengthy code.

### What is a Function?

A function is a **reusable block of code** that performs a task.

Instead of rewriting the same code again and again, you define the function once, and reuse it whenever needed.

### Why use functions?

- makes code cleaner
- avoids repetition
- easier to debug and understand  

This is how a function goes:
<br>
input --> process --> output

````python
# create a function
def greet(name):
    return "Hello " + name

print(greet("Crystal")) #prints the return value of Hello Crystal (calling a function and printing the result!)

resulted_name = greet("EOS-Programme")
print(resulted_name)
````

This is how a basic function syntax will look like. Take note that a **parameter** is a placeholder variable in the function, while the **argument** is the actual value passed in when calling your function.
<br></br>
Let's look at what other possible functions there can be!

````python
#addition
def add(a, b): #2 parameters!!
    return a + b

print(add(3, 5)) #2 arguments!!


# this function checks if a number is even
def check_even(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(check_even(5))
````

## Section 7: Conditionals (if, elif, else)

Conditionals let our program make decisions based on certain rules.

````python
temperature = 35

if temperature > 30:
    print("It is warm outsideeee, as always...")
elif temperature > 20:
    print("rainy monsoon season must be cold and refreshing!")
else:
    print("Singapore so cold wan meh?")

#ideal
````

###NOTE: the sequence of your conditions matters! always place the more restrictive condition first.

````python
temperature = 35

if temperature > 20:
    print("rainy monsoon season must be cold and refreshing!")
elif temperature > 30:
    print("It is warm outsideeee, as always...")
else:
    print("Singapore so cold wan meh?")

#this is not ideal!!
````

## Section 8: Loops

Loops let us repeat code multiple times without writing it over and over.
We can use `for` loops to repeat actions a specific number of times.

````python
# Repeating 3 times
for i in range(3):
    print("Hisssss!")
````

We can also use `while` loops to achieve the same repeeeattttinggg purpose!

````python
counter = 0

while counter < 3:
    print("Hisssss!")
    counter += 1
````

Be careful! always include a terminating condition, if not your `while` loop will constantly be running

````python
while True:
    print("I am an infinite loop!")

# no worries, just pause this cell, um your computer should be safe!
````

## Section 9: The Random Module

Python has pre-made toolboxes called "modules" that we can import.
The `random` module helps us generate random choices, like flipping a coin!

As with all python modules, always `import` your module first!

````python
import random #run this code first!!
````

````python
# Roll a 6-sided die
die_roll = random.randint(1, 6)
print("You rolled a:", die_roll)
````

Other than plain numbers, you can randomise through an Array (or List) too!

````python
colours = ["red", "blue", "green", "yellow"]
print(random.choice(colours))

random.shuffle(colours)
print(colours)
````
